```python
# Investigate payments greater than R$5,000

high_value_payments = payments[
    payments["payment_value"] > 5000
]

print("Payments above R$5000:", len(high_value_payments))

high_value_payments[
    ["order_id", "payment_type", "payment_value"]
].sort_values(
    by="payment_value",
    ascending=False
)

```

```python
# Select the order with the highest payment value

highest_order_id = payments.loc[
    payments["payment_value"].idxmax(),
    "order_id"
]

# Calculate total product price and freight for that order

order_details = order_items[
    order_items["order_id"] == highest_order_id
]

print("Order ID:", highest_order_id)
print("Total product price:", order_details["price"].sum())
print("Total freight:", order_details["freight_value"].sum())

# Calculate total payments for the same order

total_payment = payments.loc[
    payments["order_id"] == highest_order_id,
    "payment_value"
].sum()

print("Total payment:", total_payment)
```

```python
# Get order IDs associated with payments above R$5,000

high_value_order_ids = high_value_payments["order_id"].unique()

# Calculate product and freight totals for each order
item_totals = (
    order_items[
        order_items["order_id"].isin(high_value_order_ids)
    ]
    .groupby("order_id")[["price", "freight_value"]]
    .sum()
)

# Calculate total payments for each order

payment_totals = (
    payments[
        payments["order_id"].isin(high_value_order_ids)
    ]
    .groupby("order_id")["payment_value"]
    .sum()
)

# Combine totals and calculate the difference

validation = item_totals.join(payment_totals)

validation["expected_total"] = (
    validation["price"] + validation["freight_value"]
)

validation["difference"] = (
    validation["payment_value"] - validation["expected_total"]
).round(2)

validation
```

```python
# Calculate product and freight totals for every order

all_item_totals = (
    order_items.groupby("order_id")[["price", "freight_value"]]
    .sum()
)

# Calculate total payments for every order

all_payment_totals = (
    payments.groupby("order_id")["payment_value"]
    .sum()
)

# Compare only orders present in both tables

payment_validation = all_item_totals.join(
    all_payment_totals,
    how="inner"
)

payment_validation["expected_total"] = (
    payment_validation["price"]
    + payment_validation["freight_value"]
)

# Round to avoid floating-point precision issues

payment_validation["difference"] = (
    payment_validation["payment_value"]
    - payment_validation["expected_total"]
).round(2)

print("Orders compared:", len(payment_validation))

print(
    "Orders with payment discrepancies:",
    (payment_validation["difference"] != 0).sum()
)
```

```python
# Examine payment discrepancies

discrepancies = payment_validation[
    payment_validation["difference"] != 0
]

print("Discrepancy summary:")
print(discrepancies["difference"].describe())

print("\nLargest absolute discrepancies:")

discrepancies.reindex(
    discrepancies["difference"].abs()
    .sort_values(ascending=False)
    .index
).head(10)
```

```python
# Count small and large payment discrepancies

small_discrepancies = (discrepancies["difference"].abs() <= 1).sum()

large_discrepancies = (discrepancies["difference"].abs() > 1).sum()

print("Small discrepancies:", small_discrepancies)
print("Large discrepancies:", large_discrepancies)
```

```python
# Check whether order date values can be converted to datetime

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    converted_dates = pd.to_datetime(orders[column], errors="coerce")

    invalid_dates = (
        orders[column].notna() & converted_dates.isna()
    ).sum()

    print(column, "- Invalid dates:", invalid_dates)
```

```python
# Convert dates temporarily for logical validation

purchase_dates = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

delivery_dates = pd.to_datetime(
    orders["order_delivered_customer_date"]
)

# Check deliveries occurring before purchase

invalid_delivery_sequence = (
    delivery_dates < purchase_dates
).sum()

print(
    "Orders delivered before purchase:",
    invalid_delivery_sequence
)
```

```python
# Convert estimated delivery dates

estimated_dates = pd.to_datetime(
    orders["order_estimated_delivery_date"]
)

# Count orders delivered after the estimated date

late_deliveries = (
    delivery_dates.dt.normalize() > estimated_dates.dt.normalize()
).sum()

print("Orders delivered late:", late_deliveries)
```

```python
# Count orders with valid actual and estimated delivery dates

valid_delivery_orders = (
    delivery_dates.notna() & estimated_dates.notna()
).sum()

# Calculate late delivery percentage

late_delivery_percentage = (
    late_deliveries / valid_delivery_orders
) * 100

print("Orders with valid delivery dates:", valid_delivery_orders)
print("Late delivery percentage:", round(late_delivery_percentage, 2), "%")
```

```python
# Calculate delivery duration in days

delivery_duration = (
    delivery_dates - purchase_dates
).dt.total_seconds() / 86400

# Summarize delivery duration

print(delivery_duration.describe())
```

```python
# Identify deliveries taking more than 90 days

long_deliveries = delivery_duration > 90

print(
    "Orders taking more than 90 days:",
    long_deliveries.sum()
)

# Examine the longest delivery durations

print(
    "\nTop 10 longest deliveries (days):"
)

print(
    delivery_duration.sort_values(
        ascending=False
    ).head(10).round(2)
)
```

```python
# Check order statuses for deliveries longer than 90 days

print(
    orders.loc[
        long_deliveries,
        "order_status"
    ].value_counts()
)
```
